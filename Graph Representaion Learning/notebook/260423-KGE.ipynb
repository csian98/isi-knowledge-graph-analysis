# Knowledge Graph Embedding

```python
import os, sys
from pathlib import Path
from typing import List
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from tqdm import tqdm

DATA_PATH = Path("../data/FB15k-237/")
DATA_FILE = ["train.txt", "valid.txt", "test.txt"]
```

```python
def get_triplets(path: Path) -> List:
    tsv = pd.read_csv(path, header=None, sep='\t')
    return list(tsv.itertuples(index=False, name=None))

train_triplets = get_triplets(DATA_PATH / DATA_FILE[0])
valid_triplets = get_triplets(DATA_PATH / DATA_FILE[1])
test_triplets = get_triplets(DATA_PATH / DATA_FILE[2])

triplets = train_triplets + valid_triplets + test_triplets
entity2id = {}
relation2id = {}

for h, r, t in triplets:
    entity2id.setdefault(h, len(entity2id))
    entity2id.setdefault(t, len(entity2id))
    relation2id.setdefault(r, len(relation2id))

id2entity = {v:k for k, v in entity2id.items()}
id2relation = {v:k for k, v in relation2id.items()}
num_entities = len(id2entity)
num_relations = len(id2relation)

print(f"Train Triplets: {len(train_triplets)}")
print(f"Valid Triplets: {len(valid_triplets)}")
print(f"Test Triplets: {len(test_triplets)}")
print(f"# of Entities: {num_entities}")
print(f"# of Relations: {num_relations}")
```

```python
class KGDataset(Dataset):
    def __init__(self, triplets, entity2id, relation2id):
        self.data = [
            (entity2id[h], relation2id[r], entity2id[t])
            for h, r, t in triplets
        ]

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        return torch.tensor(self.data[idx], dtype=torch.long)

train_datasets = KGDataset(train_triplets, entity2id, relation2id)
train_loader = DataLoader(train_datasets, batch_size=1024, shuffle=True)
```

## Translational Distance Method

### TransE

```python
class TransE(nn.Module):
    def __init__(self, num_entities, num_relations, emb_dim=100, margin=1.0, norm=1):
        super().__init__()
        self.emb_dim = emb_dim
        self.margin = margin
        self.norm = norm

        self.entity_emb = nn.Embedding(num_entities, emb_dim)
        self.relation_emb = nn.Embedding(num_relations, emb_dim)

        nn.init.uniform_(self.entity_emb.weight, -6/emb_dim**0.5, 6/emb_dim**0.5)
        nn.init.uniform_(self.relation_emb.weight, -6/emb_dim**0.5, 6/emb_dim**0.5)

        with torch.no_grad():
            self.relation_emb.weight.data = nn.functional.normalize(
                self.relation_emb.weight.data, p=2, dim=1
            )

    def score(self, h, r, t):
        h_emb = nn.functional.normalize(self.entity_emb(h), p=2, dim=1)
        r_emb = self.relation_emb(r)
        t_emb = nn.functional.normalize(self.entity_emb(t), p=2, dim=1)

        return torch.norm(h_emb + r_emb - t_emb, p=self.norm, dim=1)

    def forward(self, pos_batch, neg_batch):
        pos_score = self.score(pos_batch[:,0], pos_batch[:,1], pos_batch[:,2])
        neg_score = self.score(neg_batch[:,0], neg_batch[:,1], neg_batch[:,2])
        loss = torch.relu(self.margin + pos_score - neg_score).mean()
        return loss
```

```python
def corrupt_batch(batch, num_entities, device):
    neg = batch.clone()
    corrupt_head = torch.rand(len(batch)) < 0.5
    rand_entities = torch.randint(0, num_entities, (len(batch), ), device=device)

    neg[corrupt_head, 0] = rand_entities[corrupt_head]
    neg[~corrupt_head, 2] = rand_entities[~corrupt_head]
    return neg
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = TransE(num_entities, num_relations, emb_dim=100, margin=1.0).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=100, gamma=0.5)

EPOCHS = 100
for epoch in range(1, EPOCHS+1):
    model.train()
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch}", leave=True)

    for batch in pbar:
        pos_batch = batch.to(device)
        neg_batch = corrupt_batch(pos_batch, num_entities, device)

        optimizer.zero_grad()
        loss = model(pos_batch, neg_batch)
        loss.backward()
        optimizer.step()

        with torch.no_grad():
            model.entity_emb.weight.data = nn.functional.normalize(
                model.entity_emb.weight.data, p=2, dim=1
            )

        total_loss += loss.item()
        pbar.set_postfix(loss=loss.item())

    avg_loss = total_loss / len(train_loader)
    scheduler.step()
    pbar.set_postfix(loss=loss.item())

def evaluate(model, triplets, all_triplets, entity2id, relation2id,
             num_entities, device, k_list=[1, 3, 10]):
    model.eval()
    ranks = []

    true_tails = {}
    for h, r, t in all_triplets:
        h_id = entity2id[h]; r_id = relation2id[r]; t_id = entity2id[t]
        true_tails.setdefault((h_id, r_id), set()).add(t_id)

    eval_data = [(entity2id[h], relation2id[r], entity2id[t]) for h, r, t in triplets]
    all_entities = torch.arange(num_entities, device=device)

    with torch.no_grad():
        for h_id, r_id, t_id in tqdm(eval_data, desc="Evaluating"):
            h_emb = nn.functional.normalize(model.entity_emb(torch.tensor([h_id], device=device)), p=2, dim=1)
            r_emb = model.relation_emb(torch.tensor([r_id], device=device))
            t_emb = nn.functional.normalize(model.entity_emb(all_entities), p=2, dim=1)
            scores = torch.norm(h_emb + r_emb - t_emb, p=model.norm, dim=1)

            filter_mask = torch.zeros(num_entities, dtype=torch.bool, device=device)
            for true_t in true_tails.get((h_id, r_id), set()):
                if true_t != t_id:
                    filter_mask[true_t] = True
            scores[filter_mask] = float('inf')

            correct_score = scores[t_id]
            rank = (scores < correct_score).sum().item() + 1
            ranks.append(rank)

    ranks = np.array(ranks)
    mrr = np.mean(1.0 / ranks)
    print(f"MRR: {mrr:.4f}")
    for k in k_list:
        hits = np.mean(ranks <= k)
        print(f"Hits@{k}: {hits:.4f}")

evaluate(model, valid_triplets, triplets, entity2id, relation2id, num_entities, device)
```

### TransH

```python
class TransH(nn.Module):
    def __init__(self, num_entities, num_relations,
                 emb_dim=100, margin=1.0, norm=1, C=0.25, eps=1e-3):
        super().__init__()
        self.emb_dim = emb_dim
        self.margin = margin
        self.norm = norm
        self.C = C
        self.eps = eps

        self.entity_emb = nn.Embedding(num_entities, emb_dim)
        self.relation_emb = nn.Embedding(num_relations, emb_dim)
        self.normal_emb = nn.Embedding(num_relations, emb_dim)

        bound = 6 / emb_dim ** 0.5
        nn.init.uniform_(self.entity_emb.weight, -bound, bound)
        nn.init.uniform_(self.relation_emb.weight, -bound, bound)
        nn.init.uniform_(self.normal_emb.weight, -bound, bound)

        with torch.no_grad():
            self.normal_emb.weight.data = nn.functional.normalize(
                self.normal_emb.weight.data, p=2, dim=1
            )

    def _project(self, emb, w_r):
        return emb - (emb * w_r).sum(dim=1, keepdim=True) * w_r

    def score(self, h, r, t):
        h_emb = self.entity_emb(h)
        r_emb = self.relation_emb(r)
        t_emb = self.entity_emb(t)
        w_r = nn.functional.normalize(self.normal_emb(r), p=2, dim=1)

        h_proj = self._project(h_emb, w_r)
        t_proj = self._project(t_emb, w_r)

        return torch.norm(h_proj + r_emb - t_proj, p=self.norm, dim=1)

    def soft_constraints(self):
        entity_norm = torch.norm(self.entity_emb.weight, p=2, dim=1)
        e_penalty = torch.relu(entity_norm ** 2 - 1.0).mean()

        w = nn.functional.normalize(self.normal_emb.weight, p=2, dim=1)
        d = self.relation_emb.weight

        ortho = (w * d).sum(dim = 1) ** 2 / (torch.norm(d, p=2, dim=1) ** 2 + 1e-8)
        o_penalty = torch.relu(ortho - self.eps ** 2).mean()

        return self.C * (e_penalty + o_penalty)

    def forward(self, pos_batch, neg_batch):
        pos_score = self.score(pos_batch[:, 0], pos_batch[:, 1], pos_batch[:, 2])
        neg_score = self.score(neg_batch[:, 0], neg_batch[:, 1], neg_batch[:, 2])
        margin_loss = torch.relu(self.margin + pos_score - neg_score).mean()
        return margin_loss + self.soft_constraints()
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = TransH(num_entities, num_relations, emb_dim=100, margin=1.0, C=0.25).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=100, gamma=0.5)

EPOCHS = 100
for epoch in range(1, EPOCHS+1):
    model.train()
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch}", leave=True)

    for batch in pbar:
        pos_batch = batch.to(device)
        neg_batch = corrupt_batch(pos_batch, num_entities, device)

        optimizer.zero_grad()
        loss = model(pos_batch, neg_batch)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        pbar.set_postfix(loss=loss.item())

    avg_loss = total_loss / len(train_loader)
    scheduler.step()
    pbar.set_postfix(loss=loss.item())

def evaluate(model, triplets, all_triplets, entity2id, relation2id,
             num_entities, device, k_list=[1, 3, 10]):
    model.eval()
    ranks = []

    true_tails = {}
    for h, r, t in all_triplets:
        h_id = entity2id[h]; r_id = relation2id[r]; t_id = entity2id[t]
        true_tails.setdefault((h_id, r_id), set()).add(t_id)

    eval_data = [(entity2id[h], relation2id[r], entity2id[t]) for h, r, t in triplets]
    all_entities = torch.arange(num_entities, device=device)

    with torch.no_grad():
        for h_id, r_id, t_id in tqdm(eval_data, desc="Evaluating"):
            h_emb = model.entity_emb(torch.tensor([h_id], device=device))
            r_emb = model.relation_emb(torch.tensor([r_id], device=device))
            t_emb = model.entity_emb(all_entities)
            w_r = nn.functional.normalize(
                model.normal_emb(torch.tensor([r_id], device=device)), p=2, dim=1
            )

            h_proj = h_emb - (h_emb * w_r).sum(dim=1, keepdim=True) * w_r
            t_proj = t_emb - (t_emb * w_r).sum(dim=1, keepdim=True) * w_r
            
            scores = torch.norm(h_proj + r_emb - t_proj, p=model.norm, dim=1)

            filter_mask = torch.zeros(num_entities, dtype=torch.bool, device=device)
            for true_t in true_tails.get((h_id, r_id), set()):
                if true_t != t_id:
                    filter_mask[true_t] = True

            scores[filter_mask] = float("inf")
            correct_score = scores[t_id]
            rank = (scores < correct_score).sum().item() + 1
            ranks.append(rank)
    
    ranks = np.array(ranks)
    mrr = np.mean(1.0 / ranks)
    print(f"MRR: {mrr:.4f}")
    for k in k_list:
        hits = np.mean(ranks <= k)
        print(f"Hits@{k}: {hits:.4f}")

evaluate(model, valid_triplets, triplets, entity2id, relation2id, num_entities, device)
```

## Semantic Matching Models

### RESCAL

```python
class RESCAL(nn.Module):
    def __init__(self, num_entities, num_relations, emb_dim=100, margin=1.0):
        super().__init__()
        self.emb_dim = emb_dim
        self.margin = margin

        self.entity_emb = nn.Embedding(num_entities, emb_dim)
        self.relation_mat = nn.Embedding(num_relations, emb_dim * emb_dim)

        nn.init.xavier_uniform_(self.entity_emb.weight)
        nn.init.xavier_uniform_(self.relation_mat.weight)

    def score(self, h, r, t):
        h_emb = self.entity_emb(h)
        M_r = self.relation_mat(r).view(-1, self.emb_dim, self.emb_dim)
        t_emb = self.entity_emb(t)

        M_t = torch.bmm(M_r, t_emb.unsqueeze(2)).squeeze(2)
        return (h_emb * M_t).sum(dim=1)

    def forward(self, pos_batch, neg_batch):
        pos_score = self.score(pos_batch[:, 0], pos_batch[:, 1], pos_batch[:, 2])
        neg_score = self.score(neg_batch[:, 0], neg_batch[:, 1], neg_batch[:, 2])

        loss = torch.relu(self.margin - pos_score + neg_score).mean()
        return loss
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = RESCAL(num_entities, num_relations, emb_dim=50, margin=1.0).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-5)
scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=100, gamma=0.5)

EPOCHS = 100

for epoch in range(1, EPOCHS + 1):
    model.train()
    total_loss = 0
    pbar = tqdm(train_loader, desc=f"Epoch {epoch}", leave=True)
 
    for batch in pbar:
        pos_batch = batch.to(device)
        neg_batch = corrupt_batch(pos_batch, num_entities, device)
 
        optimizer.zero_grad()
        loss = model(pos_batch, neg_batch)
        loss.backward()
        optimizer.step()
 
        total_loss += loss.item()
        pbar.set_postfix(loss=loss.item())
 
    avg_loss = total_loss / len(train_loader)
    scheduler.step()
    pbar.set_postfix(loss=loss.item())

def evaluate(model, triplets, all_triplets, entity2id, relation2id,
             num_entities, device, k_list=[1, 3, 10]):
    model.eval()
    ranks = []
 
    true_tails = {}
    for h, r, t in all_triplets:
        h_id = entity2id[h]; r_id = relation2id[r]; t_id = entity2id[t]
        true_tails.setdefault((h_id, r_id), set()).add(t_id)
 
    eval_data = [(entity2id[h], relation2id[r], entity2id[t]) for h, r, t in triplets]
    all_entities = torch.arange(num_entities, device=device)
 
    with torch.no_grad():
        for h_id, r_id, t_id in tqdm(eval_data, desc="Evaluating"):
            h_emb = model.entity_emb(torch.tensor([h_id], device=device))
            t_emb = model.entity_emb(all_entities)
            M_r = model.relation_mat(
                torch.tensor([r_id], device=device)
            ).view(model.emb_dim, model.emb_dim)
 
            hM = h_emb @ M_r
            scores = (hM * t_emb).sum(dim=1)
 
            filter_mask = torch.zeros(num_entities, dtype=torch.bool, device=device)
            for true_t in true_tails.get((h_id, r_id), set()):
                if true_t != t_id:
                    filter_mask[true_t] = True
            scores[filter_mask] = float('-inf')
 
            correct_score = scores[t_id]
            rank = (scores > correct_score).sum().item() + 1
            ranks.append(rank)
 
    ranks = np.array(ranks)
    mrr = np.mean(1.0 / ranks)
    print(f"MRR: {mrr:.4f}")
    for k in k_list:
        hits = np.mean(ranks <= k)
        print(f"Hits@{k}: {hits:.4f}")
 
 
evaluate(model, valid_triplets, triplets, entity2id, relation2id, num_entities, device)
```
